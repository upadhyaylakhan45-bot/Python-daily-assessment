```python

even = 0
odd = 0

for i in range(1, 51):
    if i % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even numbers =", even)
print("Odd numbers =", odd)
```

```python

data = [22, 33, 12, 32, 45, 13]

for i in range(len(data) - 1, -1, -1):
    print(data[i], end=" ")
```

```python

data = [22, 55, 12, 89, 34, 67]

largest = data[0]

for i in data:
    if i > largest:
        largest = i

print("Largest number =", largest)
```

```python

data = [22, 55, 12, 89, 34, 67]

smallest = data[0]

for i in data:
    if i < smallest:
        smallest = i

print("Smallest number =", smallest)
```

```python

text = input("Enter a string: ")
count = 0

for ch in text.lower():
    if ch in "aeiou":
        count += 1

print("Total vowels =", count)
```

```python

text = "Python is easy to learn"
count = 0

for ch in text:
    if ch == " ":
        count += 1

print("Total spaces =", count)
```
